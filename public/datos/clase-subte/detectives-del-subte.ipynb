### Evaluación de código

**a.** Debería devolver el producto escalar de dos vectores, o `None` si no tienen la misma longitud. Encontrá el error y corregilo.

````python
def calcular_producto_escalar(vector1, vector2):
    if len(vector1) != len(vector1):
        return None

    suma = 0
    for i in range(len(vector1)):
        suma += vector1[i] * vector2[i]
    return suma
````

**b. Mejorá con numpy (matriz identidad).** Esta función funciona, pero tiene 10 líneas. ¿Existe en numpy algo que la reemplace?

````python
def obtener_matriz_identidad(tamaño):
    matriz = []
    for i in range(tamaño):
        fila = []
        for j in range(tamaño):
            if i == j:
                fila.append(1)
            else:
                fila.append(0)
        matriz.append(fila)
    return matriz
````
