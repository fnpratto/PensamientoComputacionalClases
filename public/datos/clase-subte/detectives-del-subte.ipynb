## Producto final · ¿Qué día fue el más raro del año y por qué?

Encontrá el día que más se aleja del total diario promedio, graficá la serie y escribí la
conclusión: ¿es un pico hacia arriba o hacia abajo? ¿feriado, evento, paro, o un problema de
los datos? **Pista:** comparar un domingo con un martes no es justo; mirá solo los días de
semana y van a aparecer los feriados (Año Nuevo, Navidad, Carnaval, 1º de mayo...).

```python
df = pd.read_csv('subte_diario_por_linea.csv')
diario = df.groupby('fecha')['pasajeros'].sum()

media = diario.mean()
raro = (diario - media).abs().idxmax()
print('El día más raro fue', raro)

# graficá la serie y escribí tu conclusión acá abajo
```
