````python
def aprobo_parcial(nota):
		return nota >= 4
````

Para facilitar el trabajo futuro, deciden adelantarse y hacer algo similar para los exámenes finales. En este caso sólo pueden rendir final aquellos alumnos que hayan aprobado ambos parciales pero cuyo promedio sea menor a 7. Los docentes solicitan ahora un programa que dada las notas de ambos parciales devuelva True si deben/pueden rendir final y False en caso contrario.

````python
def aprobo_parciales(nota_primer_parcial, nota_segundo_parcial):
		return aprobo_parcial(nota_primer_parcial) and aprobo_parcial(nota_segundo_parcial)

def promociono(nota_primer_parcial, nota_segundo_parcial):
		return nota_primer_parcial + nota_segundo_parcial > 13

def rinde_final(nota_primer_parcial, nota_segundo_parcial):
		return aprobo_parciales(nota_primer_parcial, nota_segundo_parcial) and
		not promociono(nota_primer_parcial, nota_segundo_parcial)
````

- Si no aprobó ambos parciales el estado de la cursada es `Desaprobado`
- Si aprobó ambos parciales con promedio menor a 7 el estado de la cursada es `En final`
- Si aprobó ambos parciales con promedio mayor o igual a 7 el estado de la cursada es `Promoción`.
