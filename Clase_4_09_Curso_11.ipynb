## Sección 1 — Cadenas de caracteres

###  Recordatorio de teoría

- Se puede acceder a cada caracter por índice: `cadena[i]`.
- **Slicing**: `cadena[inicio:fin]`, y `cadena[::-1]` invierte la cadena.
- Métodos útiles: `len()`, `split()`, `split(separador)`, `join()`, `replace(a, b)`, `upper()`, `lower()`, `find(sub)` (devuelve `-1` si no está), `index(sub)` (tira error si no está), `count(sub)`.
- `find` acepta un segundo parámetro opcional: la posición desde donde empezar a buscar. `"hola".find("o", 2)`.

## Sección 2 — Rangos

### Recordatorio de teoría
- `range(inicio, fin, paso)`: el `inicio` es opcional (default 0), el `paso` es opcional (default 1).
- `range(10, 51, 5)` → 10, 15, 20, ..., 50 (el `fin` **no** se incluye, por eso se pone 51 si querés llegar a 50).
- El paso puede ser negativo, para recorrer en orden decreciente: `range(40, 19, -2)`.
- Un `range` **no es una lista**, pero se puede iterar con `for`.
- `zip(secuencia1, secuencia2)` permite recorrer dos secuencias en simultáneo, de a pares.

## Sección 3 — Tuplas

###  Recordatorio de teoría
- Las tuplas son secuencias **inmutables**: no se pueden modificar sus elementos una vez creadas. Si hay que "cambiar" algo, hay que **armar una tupla nueva**.
- Se definen con paréntesis: `(1, 2, 3)`. Una tupla de un solo elemento necesita la coma: `(1,)`.
- Se accede por índice igual que en las cadenas: `tupla[0]`.
- **Slicing**: `tupla[0:2]` devuelve una nueva tupla con esos elementos.
- **Empaquetado**: `d = a, b, c` arma una tupla.
- **Desempaquetado**: `a, b, c = d` reparte los elementos de la tupla `d` en tres variables. La cantidad de variables tiene que coincidir con la longitud de la tupla, o da error.
- `len(tupla)` devuelve la cantidad de elementos.
